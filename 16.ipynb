```python
# Practical 16 – Titanic Dataset Fare Distribution Histogram
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# -------------------------------------------
# 1. Load Titanic Dataset
# -------------------------------------------
df = pd.read_csv("Titanic.csv")   # your uploaded dataset

print("=== Titanic Dataset Preview ===")
print(df.head())

# -------------------------------------------
# 2. Plot Histogram for Fare
# -------------------------------------------
plt.figure(figsize=(8,5))
sns.histplot(df["Fare"], bins=30, kde=True, color="blue")
plt.title("Distribution of Ticket Fare on Titanic")
plt.xlabel("Fare Amount")
plt.ylabel("Number of Passengers")
plt.show()

```
