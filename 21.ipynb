```python
# K-means on IRIS dataset (K=4) — short exam-ready code
import numpy as np
import pandas as pd

# 1) Load dataset
df = pd.read_csv("IRIS.csv")
points = df.iloc[:, 0:4].values   # numeric features only (shape: N x 4)

K = 4
np.random.seed(1)

# 2) Randomly pick K points from data as initial centroids
initial_idx = np.random.choice(points.shape[0], K, replace=False)
centroids = points[initial_idx]

print("Initial Centroids:\n", centroids)

# 3) Euclidean distance function
def distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

# 4) Run K-means for 10 iterations
for it in range(10):
    # a) create empty cluster lists
    clusters = [[] for _ in range(K)]

    # b) assignment step: assign each point to nearest centroid
    for p in points:
        dists = [distance(p, centroids[c]) for c in range(K)]
        cluster_id = np.argmin(dists)
        clusters[cluster_id].append(p)

    # c) update step: recompute centroids as mean of assigned points
    for c in range(K):
        centroids[c] = np.mean(clusters[c], axis=0)

    print(f"Iteration {it+1} done")

# 5) Print final cluster means
print("\nFINAL CLUSTER CENTROIDS:")
for c in range(K):
    print(f"Cluster {c+1} centroid:", centroids[c])

```
