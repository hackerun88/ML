```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("IRIS.csv")
```

```python
print("=== Iris Dataset Preview===")
print(df.head())
      
```

```python
print("\n=== Features and their Data Types ===")
for col in df.columns:
    print(f"{col} --> {df[col].dtype}")
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
nominal_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
print("\nNumeric Features:", numeric_cols)
print("\nNominal Features:", nominal_cols)
```

```python
df.hist(figsize=(12, 8), bins=20)
plt.suptitle("Histograms for Iris Dataset FEatures", fontsize=16)
plt.tight_layout()
plt.show()
```
