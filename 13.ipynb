```python
import pandas as pd
```

```python
df = pd.read_csv("Covid Vaccine Statewise.csv")
```

```python
print("===Dataset Preview ===")
print(df.head())
```

```python
print("\n=== Dataset Information ===")
print(df.info())
```

```python
print("\n=== Statistical Description ===")
print(df.describe(include="all"))
```

```python
print("\n=== State-wise First Dose Vaccination Count ===")
first_dose = df.groupby("State")["First Dose Administered"].sum()
print(first_dose)
```

```python
print("\n=== State-wise Second Dose Vaccination Count ===")
second_dose = df.groupby("State")["Second Dose Administered"].sum()
print(second_dose)
```
