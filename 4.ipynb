```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Cosmatic.csv")
```

```python
def entropy(col):
    values, counts = np.unique(col, return_counts=true)
    probs = counts / counts.sum()
    return -np.sum(probs * np.log2(probs))
```

```python
def information_gain(df, feature, target="Buys"):
    total_entropy = entropy(df[target])
    values, counts, = np.unique(df[feature], return_counts=True)

    weighted_entropy = 0
    for v, c in zip(values, counts):
        subset = df[df[feature] == v]
        weighted_entropy += (c / len(df)) * entropy(subset[target])

    return total_entropy - weighted_entropy
```

```python
features = ["Age", "Income", "Gender", "Ms"]
ig_results = {}

print("=== Information Gain for Each Feature ===")
for f in features:
    ig = information_gain(df, f)
    ig_results[f] = ig
    print(f"{f}: {ig}")
```

```python
root = max(ig_results, key=ig_results.get)

print("\n=== ROOT NODE OF DECISION TREE ===")
print("Root Attribute:", root)
print("Information Gain:", ig_results[root])
```
