```python
# Practical 2: Summary Statistics on Telecom_Churn Dataset

import pandas as pd

# -------------------------------------------
# 1. Load Dataset
# -------------------------------------------
df = pd.read_csv("Telecom Churn.csv")
print("=== First 5 rows ===")
print(df.head())

# -------------------------------------------
# 2. MINIMUM VALUE
# -------------------------------------------
print("\n=== Minimum value of each feature ===")
print(df.min(numeric_only=True))

# -------------------------------------------
# 3. MAXIMUM VALUE
# -------------------------------------------
print("\n=== Maximum value of each feature ===")
print(df.max(numeric_only=True))

# -------------------------------------------
# 4. MEAN
# -------------------------------------------
print("\n=== Mean of each feature ===")
print(df.mean(numeric_only=True))

# -------------------------------------------
# 5. RANGE (MAX - MIN)
# -------------------------------------------
numeric_df = df.select_dtypes(include=['number'])
print("\n=== Range of each feature (max - min) ===")
print(numeric_df.max() - numeric_df.min())

# -------------------------------------------
# 6. STANDARD DEVIATION
# -------------------------------------------
print("\n=== Standard Deviation of each feature ===")
print(df.std(numeric_only=True))

# -------------------------------------------
# 7. VARIANCE
# -------------------------------------------
print("\n=== Variance of each feature ===")
print(df.var(numeric_only=True))

# -------------------------------------------
# 8. PERCENTILES (25th, 50th, 75th)
# -------------------------------------------
numeric_df = df.select_dtypes(include='number')

print("\n=== 25th Percentile ===")
print(numeric_df.quantile(0.25))

print("\n=== 50th Percentile (Median) ===")
print(numeric_df.quantile(0.50))

print("\n=== 75th Percentile ===")
print(numeric_df.quantile(0.75))

print("\n=== Percentiles (25, 50, 75) ===")
print(numeric_df.quantile([0.25, 0.50, 0.75]))
```
