```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("iris.csv")

print("First 5 rows of the dataset:")
print(df.head())

# Numeric columns only (exclude species)
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

# Unique species available in dataset
species_list = df['species'].unique()

# Function to compute detailed statistics for each species
def detailed_stats(species):
    print("\n==============================================")
    print(f"   Statistical Summary for: {species}")
    print("==============================================")
    
    # Filter dataset for the given species
    data = df[df['species'] == species][numeric_cols]
    
    # Create custom summary
    summary = pd.DataFrame({
        "Mean": data.mean(),
        "Median": data.median(),
        "Std Dev": data.std(),
        "Min": data.min(),
        "Max": data.max(),
        "5th Percentile": data.quantile(0.05),
        "25th Percentile": data.quantile(0.25),
        "50th Percentile": data.quantile(0.50),
        "75th Percentile": data.quantile(0.75),
        "95th Percentile": data.quantile(0.95)
    })
    
    print(summary)

# Compute stats for each species
for s in species_list:
    detailed_stats(s)
```
