```python
# Practical 1: Basic Data Operations on Titanic Dataset using Python

import pandas as pd

# -----------------------------
# 1. READ DATA FROM DIFFERENT FORMATS
# -----------------------------

# Read from CSV file
titanic_csv = pd.read_csv("Titanic.csv")
print("=== First 5 rows from CSV ===")
print(titanic_csv.head())

# Read from Excel file
# (Make sure titanic.xlsx exists. You can save the same CSV as Excel.)
titanic_xls = pd.read_excel("Titanic.xlsx")
print("\n=== First 5 rows from Excel ===")
print(titanic_xls.head())

# For further operations, we will use titanic_csv
df = titanic_csv

# -----------------------------
# 2. INDEXING AND SELECTING DATA
# -----------------------------

print("\n=== Shape of dataset (rows, columns) ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns)

# Select a single column (e.g. 'Name')
print("\n=== First 5 Passenger Names ===")
print(df["Name"].head())

# Select multiple columns (e.g. 'Name', 'Sex', 'Age')
print("\n=== Name, Sex, Age (first 5 rows) ===")
print(df[["Name", "Sex", "Age"]].head())

# Select rows by index using iloc (e.g. first 5 rows)
print("\n=== First 5 rows using iloc ===")
print(df.iloc[0:5])

# Select rows by label using loc (if index has labels)
# Example: show passengers with PassengerId 1 to 5
print("\n=== Rows with PassengerId from 1 to 5 using loc ===")
print(df.loc[0:4])   # if default index, 0 to 4

# Boolean indexing: passengers who survived
print("\n=== Passengers who survived (first 5) ===")
print(df[df["Survived"] == 1].head())

# Boolean indexing: passengers older than 50
print("\n=== Passengers with Age > 50 (first 5) ===")
print(df[df["Age"] > 50].head())

# -----------------------------
# 3. SORT DATA
# -----------------------------

# Sort by Age in ascending order
print("\n=== Data sorted by Age (ascending, first 5) ===")
print(df.sort_values(by="Age", ascending=True).head())

# Sort by Fare in descending order
print("\n=== Data sorted by Fare (descending, first 5) ===")
print(df.sort_values(by="Fare", ascending=False).head())

# Sort by multiple columns: Pclass then Age
print("\n=== Data sorted by Pclass, then Age (first 5) ===")
print(df.sort_values(by=["Pclass", "Age"], ascending=[True, True]).head())

# -----------------------------
# 4. DESCRIBE ATTRIBUTES OF DATA
# -----------------------------

print("\n=== Summary statistics of numerical features ===")
print(df.describe())

print("\n=== Summary statistics of all columns (including object) ===")
print(df.describe(include="all"))

# -----------------------------
# 5. CHECK DATA TYPES OF EACH COLUMN
# -----------------------------

print("\n=== Data Types of Each Column ===")
print(df.dtypes)

# More detailed info about dataset
print("\n=== Full info about dataset ===")
print(df.info())

```
