```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("House Data.csv")
```

```python
# Select numeric columns only
numeric_df = df.select_dtypes(include=['int64', 'float64'])
```

```python
# Standard Deviation
print("Standard Deviation:\n", numeric_df.std())
```

```python
# Percentiles (25%, 50%, 75%)
print("\nPercentiles:\n", numeric_df.quantile([0.25, 0.50, 0.75]))
```

```python
# Histograms for each feature
numeric_df.hist(figsize=(14, 12), bins=30) 
# This line creates a grid of histograms — 1 histogram per numeric column.
plt.suptitle("Feature Distributions - House Price Dataset", fontsize=16)
plt.tight_layout()#adjusts spacing
plt.show()
```
