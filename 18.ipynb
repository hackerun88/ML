```python
# Simple, robust grouped summary for House Data
import pandas as pd

# 1) Load dataset
df = pd.read_csv("House Data.csv")   # use your file name
print("=== DATASET PREVIEW ===")
print(df.head())
print("\n=== COLUMNS ===")
print(list(df.columns))

# 2) Choose categorical & quantitative variables (sensible defaults)
categorical_var = "district"   # from your column list
quantitative_var = "price"     # from your column list

# 3) Validate columns exist
missing = [c for c in (categorical_var, quantitative_var) if c not in df.columns]
if missing:
    print(f"\nError: column(s) not found: {missing}")
    print("Available columns (first 30):", list(df.columns)[:30])
    raise SystemExit("Fix column names and re-run the script.")

# 4) Convert quantitative column to numeric (safe) and show how many became NaN
df[quantitative_var] = pd.to_numeric(df[quantitative_var], errors="coerce")
n_missing_before = df[quantitative_var].isna().sum()
print(f"\nNote: {n_missing_before} rows in '{quantitative_var}' could not be converted to numeric (set to NaN).")

# 5) Optionally drop rows where categorical is missing or quantitative is NaN
clean_df = df.dropna(subset=[categorical_var, quantitative_var])
print(f"Rows before cleaning: {len(df)}, after cleaning: {len(clean_df)}")

# 6) Group and compute summary statistics
summary = clean_df.groupby(categorical_var)[quantitative_var].agg(
    Count='count',
    Mean='mean',
    Median='median',
    Minimum='min',
    Maximum='max',
    Std_Dev='std'
).sort_values(by='Count', ascending=False)

print(f"\n=== Summary statistics of '{quantitative_var}' grouped by '{categorical_var}' ===")
print(summary)

# 7) Save results (optional)
summary.to_csv("HousePrice_Grouped_Summary_by_district.csv")
print("\nSaved grouped summary to 'HousePrice_Grouped_Summary_by_district.csv'")
```
